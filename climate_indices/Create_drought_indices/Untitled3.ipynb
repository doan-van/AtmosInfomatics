```python
import numpy as np
import matplotlib.pyplot as plt

# Create a simple 2D dataset with a wave pattern for demonstration
x = np.linspace(0, 4 * np.pi, 100)
y = np.linspace(0, 4 * np.pi, 100)
X, Y = np.meshgrid(x, y)
data = np.sin(X) + np.cos(Y)

# Apply 2D FFT
fft_data = np.fft.fft2(data)

# Without shifting the zero frequency component
magnitude_unshifted = np.abs(fft_data)

# With shifting the zero frequency component to the center
fft_data_shifted = np.fft.fftshift(fft_data)
magnitude_shifted = np.abs(fft_data_shifted)

# Plot both the unshifted and shifted magnitude spectrums
fig, axs = plt.subplots(1, 2, figsize=(12, 5))

# Plot the unshifted FFT spectrum
axs[0].imshow(np.log1p(magnitude_unshifted), cmap='viridis')
axs[0].set_title("Unshifted FFT Spectrum")
axs[0].axis('off')

# Plot the shifted FFT spectrum
axs[1].imshow(np.log1p(magnitude_shifted), cmap='viridis')
axs[1].set_title("Shifted FFT Spectrum (Zero Centered)")
axs[1].axis('off')

plt.show()
```
