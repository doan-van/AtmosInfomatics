```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# Define a sample dataset with one-hot encoding
data = pd.DataFrame({
    'Milk': [1, 1, 1, 0, 0, 1],
    'Bread': [1, 0, 1, 1, 1, 1],
    'Butter': [0, 0, 0, 1, 0, 1],
    'Diaper': [0, 1, 0, 0, 1, 0]
})

# Step 1: Find frequent itemsets with a minimum support of 0.5
frequent_itemsets = apriori(data, min_support=0.5, use_colnames=True)

# Step 2: Generate association rules with a minimum confidence of 0.7
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7,num_itemsets=2)

# Display the association rules
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']])
```

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# Sample one-hot encoded transaction data
data = pd.DataFrame({
    'Milk': [1, 1, 1, 0, 0, 1],
    'Bread': [1, 0, 1, 1, 1, 1],
    'Butter': [0, 0, 0, 1, 0, 1],
    'Diaper': [0, 1, 0, 0, 1, 0]
})

# Function to find best support for a desired number of itemsets
def find_optimal_itemsets(data, target_num_itemsets):
    # Start with high support and reduce iteratively
    support = 0.6
    step = 0.05
    itemsets_found = 0

    # Loop to adjust support level
    while itemsets_found < target_num_itemsets and support > 0:
        frequent_itemsets = apriori(data, min_support=support, use_colnames=True)
        itemsets_found = len(frequent_itemsets)
        
        print(f"Support: {support:.2f}, Itemsets Found: {itemsets_found}")
        
        # Break if the itemset count is near the target
        if itemsets_found >= target_num_itemsets:
            break
        support -= step  # Reduce support to find more itemsets
    
    return frequent_itemsets

# Define desired number of itemsets
target_num_itemsets = 5

# Find optimal itemsets
optimal_itemsets = find_optimal_itemsets(data, target_num_itemsets)

# Display the resulting itemsets
print("\nOptimal Itemsets:")
print(optimal_itemsets)
```

```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules

# Example data where each row represents SPI categories for different regions at a given time
data = pd.DataFrame({
    'region_A': ['dry', 'dry', 'wet', 'normal', 'wet'],
    'region_B': ['wet', 'dry', 'normal', 'wet', 'dry'],
    'region_C': ['normal', 'dry', 'wet', 'normal', 'wet']
})

# Convert the categorical data to one-hot encoding
data_one_hot = pd.get_dummies(data)

# Step 1: Use Apriori to find frequent itemsets
frequent_itemsets = apriori(data_one_hot, min_support=0.2, use_colnames=True)

# Step 2: Generate association rules with a minimum confidence threshold
rules = association_rules(frequent_itemsets, metric="confidence", min_threshold=0.7)

# Display the association rules
print(rules)
```
