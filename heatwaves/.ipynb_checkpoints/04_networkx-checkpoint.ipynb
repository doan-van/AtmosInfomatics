```python
import os
import pandas as pd
pd.set_option('future.no_silent_downcasting', True)

# Load list of station pairs
stations_file = 'list_of_stations.csv'
df_stations = pd.read_csv(stations_file, index_col=0)
stids = df_stations.index[:]

# Base directory containing synchronization results
base_output_dir = 'synchronization_results'

# Initialize container for the synchronization rates
sync_rates = {}

# Iterate through each pair of stations to read synchronization rates
for point_a in stids:
    for point_b in stids:
        if point_a == point_b:
            continue

        # Define station pair directory
        station_pair = f'{point_a}_and_{point_b}'
        station_pair_dir = os.path.join(base_output_dir, station_pair)
        rates_dir = os.path.join(station_pair_dir, 'rates')

        # Read synchronization rates
        rates_file = os.path.join(rates_dir, 'synchronization_rates.txt')
        if os.path.exists(rates_file):
            with open(rates_file, 'r') as f:
                rates = f.readlines()
            sync_rates[(point_a, point_b)] = {
                'sync_rate_a_after_b_per_a': float(rates[0].strip().split(': ')[1].replace('%', '')) / 100,
                'sync_rate_b_after_a_per_b': float(rates[1].strip().split(': ')[1].replace('%', '')) / 100,
                'overall_sync_rate': float(rates[2].strip().split(': ')[1].replace('%', '')) / 100
            }

# Create DataFrames for each type of synchronization rate
sync_rate_a_after_b_df = pd.DataFrame(index=stids, columns=stids) #.fillna(0)
sync_rate_b_after_a_df = pd.DataFrame(index=stids, columns=stids) #.fillna(0)
overall_sync_rate_df = pd.DataFrame(index=stids, columns=stids) #.fillna(0)

# Fill DataFrames with synchronization rates
for (point_a, point_b), rates in sync_rates.items():
    sync_rate_a_after_b_df.loc[point_a, point_b] = rates['sync_rate_a_after_b_per_a']
    sync_rate_b_after_a_df.loc[point_a, point_b] = rates['sync_rate_b_after_a_per_b']
    overall_sync_rate_df.loc[point_a, point_b] = rates['overall_sync_rate']

# Print DataFrames
#print("Synchronization rate (A after B per A event):")
#print(sync_rate_a_after_b_df)
#print("\nSynchronization rate (B after A per B event):")
#print(sync_rate_b_after_a_df)
#print("\nOverall synchronization rate:")
#print(overall_sync_rate_df)
```

```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt

df = overall_sync_rate_df
# Assuming the DataFrame is already loaded into `df`
# If your dataframe is in a CSV or another format, you can load it like so:
# df = pd.read_csv("your_data.csv", index_col=0)

# Create a graph object
G = nx.Graph()

# Add nodes (stations)
for station in df.index:
    G.add_node(station)

# Add edges with weights from the DataFrame (excluding NaN values)
for i in df.index:
    for j in df.columns:
        if pd.notna(df.loc[i, j]) and i != j:  # Avoid self-loops
            G.add_edge(i, j, weight=df.loc[i, j])

# Step 1: Visualize the graph (basic visualization)
plt.figure(figsize=(12, 12))
pos = nx.spring_layout(G)  # Spring layout for a better visualization

# Draw the graph
nx.draw(G, pos, with_labels=True, node_size=500, node_color="lightblue", font_size=10, font_weight='bold', edge_color="gray")

# Add edge weights (the relationship values)
edge_labels = nx.get_edge_attributes(G, 'weight')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels)

plt.title("Station Relationship Graph")
plt.show()

# Optionally, if you want to plot a graph with specific thresholds (e.g., only edges with weight above 0.1):
threshold = 0.1
filtered_graph = nx.Graph()
for i in df.index:
    for j in df.columns:
        weight = df.loc[i, j]
        if pd.notna(weight) and weight > threshold and i != j:
            filtered_graph.add_edge(i, j, weight=weight)

# Visualize the filtered graph
plt.figure(figsize=(12, 12))
pos = nx.spring_layout(filtered_graph)
nx.draw(filtered_graph, pos, with_labels=True, node_size=500, node_color="lightgreen", font_size=10, font_weight='bold', edge_color="gray")
edge_labels = nx.get_edge_attributes(filtered_graph, 'weight')
nx.draw_networkx_edge_labels(filtered_graph, pos, edge_labels=edge_labels)
plt.title(f"Filtered Station Relationship Graph (Threshold > {threshold})")
plt.show()
```
