```python
import pandas as pd
from datetime import timedelta
import os

def relaxed_overlap_sync(file_a, file_b, time_window=3):
    """
    Detect synchronized events between two sets of detected events within a specified time window.
    
    Parameters:
        file_a (str): Path to the first events file.
        file_b (str): Path to the second events file.
        time_window (int): Time window in days for relaxed synchronization (default=3).
        
    Returns:
        pd.DataFrame: Synchronized events with details from both stations.
    """
    # Extract station names from file names
    station_a_name = os.path.basename(file_a).split('_')[0]
    station_b_name = os.path.basename(file_b).split('_')[0]

    # Load the event files
    station_a_events = pd.read_csv(file_a, parse_dates=['start_date', 'end_date'])
    station_b_events = pd.read_csv(file_b, parse_dates=['start_date', 'end_date'])

    # Initialize list to store synchronized events
    synchronized_events_list = []

    # Compare each event in station_a_events with all events in station_b_events
    for _, event_a in station_a_events.iterrows():
        for _, event_b in station_b_events.iterrows():
            # Check if events overlap within the time window
            # Equation: (start_date_A - time_window) <= end_date_B AND (end_date_A + time_window) >= start_date_B
            # This is a relaxed overlap method where events are considered synchronized if they
            # overlap within a specified time window of ±time_window days. This allows for some flexibility
            # in synchronization, accounting for potential uncertainties in event timing.
            # 
            # Example:
            # Suppose Station A has a heatwave event from June 1 to June 5, and Station B has an event from June 4 to June 8.
            # With a time window of ±3 days, these events are considered synchronized because they overlap within the window.
            # This flexibility is useful when events may not align perfectly due to slight differences in timing between stations.
            if (event_a['start_date'] - timedelta(days=time_window) <= event_b['end_date'] and
                event_a['end_date'] + timedelta(days=time_window) >= event_b['start_date']):
                
                synchronized_events_list.append({
                    f'{station_a_name}_start_date': event_a['start_date'],
                    f'{station_a_name}_end_date': event_a['end_date'],
                    f'{station_a_name}_duration': event_a['duration'],
                    f'{station_b_name}_start_date': event_b['start_date'],
                    f'{station_b_name}_end_date': event_b['end_date'],
                    f'{station_b_name}_duration': event_b['duration']
                })

    # Convert synchronized events to a DataFrame
    synchronized_events_df = pd.DataFrame(synchronized_events_list)

    return synchronized_events_df

# Example Usage
file_a = 'output_heatwaves/47401_temp-max_C_q0.9_d3_events.csv'
file_b = 'output_heatwaves/47520_temp-max_C_q0.9_d3_events.csv'

# Detect synchronization with a time window of ±3 days
synchronized_events_df = relaxed_overlap_sync(file_a, file_b, time_window=3)
```

```python
import pandas as pd
from datetime import timedelta
import os

def event_sequence_sync(file_a, file_b, max_lag=3):
    """
    Perform lag-based synchronization between two heatwave event files in both directions.
    
    Parameters:
        file_a (str): Path to the first heatwave file (CSV format).
        file_b (str): Path to the second heatwave file (CSV format).
        max_lag (int): Maximum lag in days for synchronization.
        
    Returns:
        dict: A dictionary containing synchronized heatwave events and synchronization rates.
    """
    # Extract station names from file names
    station_a_name = os.path.basename(file_a).split('_')[0]
    station_b_name = os.path.basename(file_b).split('_')[0]

    # Load the event files
    station_a_events = pd.read_csv(file_a, parse_dates=['start_date', 'end_date'])
    station_b_events = pd.read_csv(file_b, parse_dates=['start_date', 'end_date'])

    # Initialize lists to store synchronized events for both sequences
    events_a_after_b = []
    events_b_after_a = []

    # Iterate over events in both datasets to calculate lags
    for _, event_a in station_a_events.iterrows():
        for _, event_b in station_b_events.iterrows():
            # Calculate lag between start dates
            lag = (event_a['start_date'] - event_b['start_date']).days
            
            if 0 <= lag <= max_lag:
                # Event A occurs after Event B within the lag window
                events_a_after_b.append({
                    f'{station_a_name}_start_date': event_a['start_date'],
                    f'{station_a_name}_duration': event_a['duration'],
                    f'{station_b_name}_start_date': event_b['start_date'],
                    f'{station_b_name}_duration': event_b['duration'],
                    'lag_days': lag,
                    'sequence': f'{station_b_name}_then_{station_a_name}'
                })
            elif 0 <= -lag <= max_lag:
                # Event B occurs after Event A within the lag window
                events_b_after_a.append({
                    f'{station_b_name}_start_date': event_b['start_date'],
                    f'{station_b_name}_duration': event_b['duration'],
                    f'{station_a_name}_start_date': event_a['start_date'],
                    f'{station_a_name}_duration': event_a['duration'],
                    'lag_days': -lag,
                    'sequence': f'{station_a_name}_then_{station_b_name}'
                })

    # Convert synchronized events to DataFrames
    events_a_after_b_df = pd.DataFrame(events_a_after_b)
    events_b_after_a_df = pd.DataFrame(events_b_after_a)

    # Calculate synchronization rates
    sync_rate_a_after_b_per_a = len(events_a_after_b_df) / len(station_a_events) if len(station_a_events) > 0 else 0
    sync_rate_b_after_a_per_b = len(events_b_after_a_df) / len(station_b_events) if len(station_b_events) > 0 else 0
    overall_sync_rate = (len(events_a_after_b_df) + len(events_b_after_a_df)) / (len(station_a_events) + len(station_b_events)) if (len(station_a_events) + len(station_b_events)) > 0 else 0

    # Print synchronization rates
    #print(f"Synchronization rate for {station_a_name} occurring after {station_b_name} (per {station_a_name} event): {sync_rate_a_after_b_per_a:.2%}")
    #print(f"Synchronization rate for {station_b_name} occurring after {station_a_name} (per {station_b_name} event): {sync_rate_b_after_a_per_b:.2%}")
    print(f"Overall synchronization rate: {overall_sync_rate:.2%}")

    return {
        'events_a_after_b': events_a_after_b_df,
        'events_b_after_a': events_b_after_a_df,
        'sync_rate_a_after_b_per_a': sync_rate_a_after_b_per_a,
        'sync_rate_b_after_a_per_b': sync_rate_b_after_a_per_b,
        'overall_sync_rate': overall_sync_rate
    }

```

```python
df_stations = pd.read_csv('list_of_stations.csv', index_col=0)
stids = df_stations.index

for point_a in stids:
    print(point_a)
    for point_b in stids:
        if point_a == point_b: continue
        # Example Usage
        file_a = 'output_heatwaves/'+str(point_a)+'_temp-max_C_q0.9_d3_events.csv'
        file_b = 'output_heatwaves/'+str(point_b)+'_temp-max_C_q0.9_d3_events.csv'
        
        # Detect synchronization with a maximum lag of ±3 days
        
        # Extract station names from file names (to reuse in saving outputs)
        station_a_name = os.path.basename(file_a).split('_')[0]
        station_b_name = os.path.basename(file_b).split('_')[0]
        
        # The output consists of two DataFrames representing sequences of events and synchronization rates.
        result = event_sequence_sync(file_a, file_b, max_lag=3)
        
        # Accessing results
        events_a_after_b_df = result['events_a_after_b']
        events_b_after_a_df = result['events_b_after_a']
        sync_rate_a_after_b_per_a = result['sync_rate_a_after_b_per_a']
        sync_rate_b_after_a_per_b = result['sync_rate_b_after_a_per_b']
        overall_sync_rate = result['overall_sync_rate']
        
        # Save the results to CSV files
        output_dir = 'synchronization_results'
        os.makedirs(output_dir, exist_ok=True)
        
        events_a_after_b_df.to_csv(os.path.join(output_dir, f'{station_a_name}_after_{station_b_name}_sync_events.csv'), index=False)
        events_b_after_a_df.to_csv(os.path.join(output_dir, f'{station_b_name}_after_{station_a_name}_sync_events.csv'), index=False)
        
        # Save synchronization rates to a text file
        with open(os.path.join(output_dir, 'synchronization_rates.txt'), 'w') as f:
            f.write(f"Synchronization rate for {station_a_name} occurring after {station_b_name} (per {station_a_name} event): {sync_rate_a_after_b_per_a:.2%}\n")
            f.write(f"Synchronization rate for {station_b_name} occurring after {station_a_name} (per {station_b_name} event): {sync_rate_b_after_a_per_b:.2%}\n")
            f.write(f"Overall synchronization rate: {overall_sync_rate:.2%}\n")
```
