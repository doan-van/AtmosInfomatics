```python
import networkx as nx
import community as community_louvain
import matplotlib.pyplot as plt

# Step 1: Create a graph (you can use your own data here)
G = nx.erdos_renyi_graph(150, 0.5)  # Random graph with 30 nodes and 5% edge probability

# Step 2: Visualize the original network
plt.figure(figsize=(8, 8))

# Draw the original graph
pos = nx.spring_layout(G)  # Layout for node positions
nx.draw(G, pos, with_labels=True, node_size=500, node_color="lightblue", font_size=12)

plt.title("Original Network")
plt.show()

# Step 3: Apply the Louvain method for community detection
partition = community_louvain.best_partition(G)

# Step 4: Visualize the graph with communities
plt.figure(figsize=(8, 8))

# Draw the graph with nodes colored based on their community
nx.draw(G, pos, with_labels=True, node_size=500, node_color=list(partition.values()), cmap=plt.cm.jet, font_size=12)

plt.title("Network After Clustering (Louvain Method)")
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt

# Step 1: Create a graph (you can use your own data here)
G = nx.erdos_renyi_graph(30, 0.05)  # Random graph with 30 nodes and 5% edge probability

# Step 2: Visualize the graph
plt.figure(figsize=(8, 8))
nx.draw(G, with_labels=True, node_size=500, node_color="lightblue", font_size=12)
plt.title("Original Graph")
plt.show()

# Step 3: Degree Centrality
degree_centrality = nx.degree_centrality(G)
print("Degree Centrality:")
print(degree_centrality)

# Step 4: Betweenness Centrality
betweenness_centrality = nx.betweenness_centrality(G)
print("Betweenness Centrality:")
print(betweenness_centrality)

# Step 5: Closeness Centrality
closeness_centrality = nx.closeness_centrality(G)
print("Closeness Centrality:")
print(closeness_centrality)

# Step 6: Eigenvector Centrality
eigenvector_centrality = nx.eigenvector_centrality(G)
print("Eigenvector Centrality:")
print(eigenvector_centrality)

# Step 7: Clustering Coefficient (Local and Global)
clustering_coeff = nx.clustering(G)
print("Clustering Coefficients:")
print(clustering_coeff)

# Step 8: Average Clustering Coefficient
avg_clustering_coeff = nx.average_clustering(G)
print("Average Clustering Coefficient:")
print(avg_clustering_coeff)

# Step 9: Assortativity (Degree Assortativity)
assortativity = nx.degree_assortativity_coefficient(G)
print("Degree Assortativity Coefficient:")
print(assortativity)

# Step 10: Graph Density (Ratio of edges to possible edges)
density = nx.density(G)
print("Graph Density:")
print(density)

# Step 11: Diameter and Radius of the graph
if nx.is_connected(G):
    diameter = nx.diameter(G)
    radius = nx.radius(G)
    print("Graph Diameter:", diameter)
    print("Graph Radius:", radius)
else:
    print("The graph is not connected, so diameter and radius are not defined.")

# Step 12: Modularity (Community Detection)
# First, apply a community detection algorithm (e.g., Louvain method)
import community as community_louvain

partition = community_louvain.best_partition(G)
modularity = community_louvain.modularity(partition, G)
print("Modularity:", modularity)

# Step 13: Visualize the graph with communities (Louvain method)
plt.figure(figsize=(8, 8))
nx.draw(G, with_labels=True, node_size=500, node_color=list(partition.values()), cmap=plt.cm.jet, font_size=12)
plt.title("Graph with Communities (Louvain Method)")
plt.show()
```

```python
import networkx as nx
import matplotlib.pyplot as plt
import community as community_louvain

# Step 1: Create a graph (you can use your own data here)
G = nx.erdos_renyi_graph(30, 0.05)  # Random graph with 30 nodes and 5% edge probability

# Step 2: Compute centralities and other metrics
degree_centrality = nx.degree_centrality(G)
betweenness_centrality = nx.betweenness_centrality(G)
closeness_centrality = nx.closeness_centrality(G)
eigenvector_centrality = nx.eigenvector_centrality(G)
clustering_coeff = nx.clustering(G)
avg_clustering_coeff = nx.average_clustering(G)
assortativity = nx.degree_assortativity_coefficient(G)

# Step 3: Community Detection and Modularity
partition = community_louvain.best_partition(G)
modularity = community_louvain.modularity(partition, G)

# Step 4: Plot the metrics

# Plot Degree Centrality
plt.figure(figsize=(10, 6))
plt.bar(degree_centrality.keys(), degree_centrality.values(), color='skyblue')
plt.title('Degree Centrality')
plt.xlabel('Node')
plt.ylabel('Degree Centrality')
plt.show()

# Plot Betweenness Centrality
plt.figure(figsize=(10, 6))
plt.bar(betweenness_centrality.keys(), betweenness_centrality.values(), color='lightgreen')
plt.title('Betweenness Centrality')
plt.xlabel('Node')
plt.ylabel('Betweenness Centrality')
plt.show()

# Plot Closeness Centrality
plt.figure(figsize=(10, 6))
plt.bar(closeness_centrality.keys(), closeness_centrality.values(), color='orange')
plt.title('Closeness Centrality')
plt.xlabel('Node')
plt.ylabel('Closeness Centrality')
plt.show()

# Plot Eigenvector Centrality
plt.figure(figsize=(10, 6))
plt.bar(eigenvector_centrality.keys(), eigenvector_centrality.values(), color='salmon')
plt.title('Eigenvector Centrality')
plt.xlabel('Node')
plt.ylabel('Eigenvector Centrality')
plt.show()

# Plot Clustering Coefficients
plt.figure(figsize=(10, 6))
plt.bar(clustering_coeff.keys(), clustering_coeff.values(), color='lightcoral')
plt.title('Clustering Coefficient')
plt.xlabel('Node')
plt.ylabel('Clustering Coefficient')
plt.show()

# Plot Assortativity
plt.figure(figsize=(10, 6))
plt.bar([0], [assortativity], color='purple')
plt.title('Degree Assortativity Coefficient')
plt.ylabel('Assortativity')
plt.xticks([0], ['Graph'])
plt.show()

# Plot Average Clustering Coefficient
plt.figure(figsize=(10, 6))
plt.bar([0], [avg_clustering_coeff], color='blue')
plt.title('Average Clustering Coefficient')
plt.ylabel('Average Clustering Coefficient')
plt.xticks([0], ['Graph'])
plt.show()

# Plot Modularity (community structure)
plt.figure(figsize=(10, 6))
plt.bar([0], [modularity], color='gold')
plt.title('Modularity (Community Structure)')
plt.ylabel('Modularity')
plt.xticks([0], ['Graph'])
plt.show()

```
